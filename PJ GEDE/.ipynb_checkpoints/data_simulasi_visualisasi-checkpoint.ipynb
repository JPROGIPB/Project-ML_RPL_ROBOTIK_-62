```python
# Visualisasi episode — gaya mirip screenshot
import math
import pandas as pd
import matplotlib.pyplot as plt

# Sesuaikan nomor episode di sini
EPISODE = 99

COLS = [
    'episode','step','robot_x','robot_y','battery',
    'trash_x','trash_y','distance','progress','action',
    'reward','done','description'
]

df = pd.read_csv('rl_training_data.csv', header=None, names=COLS, encoding='utf-8')
ep = df[df['episode'] == EPISODE].sort_values('step')
if ep.empty:
    raise ValueError(f"Tidak ditemukan data untuk episode {EPISODE}")

# ambil seluruh jalur dan posisi terakhir sebagai 'current'
xs = ep['robot_x'].tolist()
ys = ep['robot_y'].tolist()
cur_x = int(xs[-1])
cur_y = int(ys[-1])
trash_x = int(ep['trash_x'].iloc[-1])
trash_y = int(ep['trash_y'].iloc[-1])
cur_step = int(ep['step'].iloc[-1])
cur_batt = ep['battery'].iloc[-1]
cur_progress = ep['progress'].iloc[-1]
cur_action_code = int(ep['action'].iloc[-1])
cur_reward = float(ep['reward'].iloc[-1])
cur_done = bool(ep['done'].iloc[-1])

# action mapping (sama seperti data_generator/dqn_agent)
ACTION_MAP = {0: 'Utara', 1: 'Selatan', 2: 'Timur', 3: 'Barat', 4: 'Kumpulkan'}
cur_action = ACTION_MAP.get(cur_action_code, str(cur_action_code))

# ukuran figure dan warna dasar
fig, ax = plt.subplots(figsize=(8, 7))
fig.patch.set_facecolor('#0f2740')
ax.set_facecolor('#0f2740')

# gaya grid mirip screenshot
ax.grid(True, color='white', linestyle='-', linewidth=0.8, alpha=0.18)

# plot jalur robot
ax.plot(xs, ys, color='#64b5ff', linewidth=1.6, marker='o', markersize=10,
        markerfacecolor='#64b5ff', markeredgecolor='none', alpha=0.95, label='Robot Path')

# titik robot sekarang (merah)
ax.scatter(cur_x, cur_y, color='#ff4d4d', s=130, zorder=5, label='Current Robot')
ax.text(cur_x + 1, cur_y - 1.5, f'Robot ({cur_x}, {cur_y})', color='white', fontsize=10)

# titik sampah (hijau)
ax.scatter(trash_x, trash_y, color='#7de27d', s=130, zorder=5, label='Nearest Trash')
ax.text(trash_x + 1, trash_y, f'({trash_x}, {trash_y})', color='white', fontsize=10)

# garis putus-putus dari robot ke sampah + anotasi jarak
ax.plot([cur_x, trash_x], [cur_y, trash_y], linestyle='--', color='#ffd24d', linewidth=2)
mid_x = (cur_x + trash_x) / 2
mid_y = (cur_y + trash_y) / 2
jarak = math.hypot(trash_x - cur_x, trash_y - cur_y)
ax.text(mid_x, mid_y, f'Jarak: {jarak:.2f}', color='#ffd24d', fontsize=11)

# batas sumbu, sesuaikan agar mirip contoh
ax.set_xlim(0, 105)
ax.set_ylim(0, 150)
ax.set_xlabel('X-Coordinate', color='white')
ax.set_ylabel('Y-Coordinate', color='white')
ax.tick_params(colors='white')

# legenda dengan latar putih kecil
leg = ax.legend(loc='upper right', frameon=True)
leg.get_frame().set_facecolor('white')
leg.get_frame().set_alpha(0.9)

# tambahkan kotak info di bawah seperti screenshot
info_lines = [
    f'Episode {EPISODE}',
    f'langkah {cur_step} : baterai {cur_batt:.1f}%: Robot di: ({cur_x}, {cur_y})',
    f'Sampah terdekat: ({trash_x}, {trash_y})',
    f'Jarak: {jarak:.2f}',
    f'Progress: {cur_progress:.1f}%',
    f'Aksi: {cur_action}  {cur_reward:.2f}',
    f'Selesai: {"Ya" if cur_done else "Tidak"}.',
]
info_text = "\n".join(info_lines)

plt.subplots_adjust(bottom=0.25)
fig.text(0.02, 0.02, info_text, fontsize=10, color='black',
         bbox=dict(facecolor='white', alpha=0.95, boxstyle='round'))

plt.show()
```
